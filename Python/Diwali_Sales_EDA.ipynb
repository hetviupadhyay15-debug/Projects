```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/Diwali Sales Data.csv",encoding='latin-1')
df
```

```python
df.head(10)
```

```python
df.shape
```

```python
df.info()
```

```python
df.drop(['Status','unnamed1'],axis=1, inplace=True, errors='ignore')
```

```python
df.info()
```

```python
pd.isnull(df)
```

```python
pd.isnull(df).sum()
```

```python
df.shape
```

```python
df.dropna(inplace=True)
df.shape
```

```python
df['Amount'] = df['Amount'].astype('int')
df['Amount'].dtypes
```

```python
df.describe()
```

```python
# Gender :-
GD = sns.countplot(x ='Gender', data = df)
plt.title("Gender Distribution")
GD.bar_label(GD.containers[0])
plt.show()
```

```python
Sales_Gen = df.groupby('Gender')['Amount'].sum().reset_index()
Sales_Gen
```

```python
plt.figure(figsize = (5,5))
plt.bar(Sales_Gen['Gender'],Sales_Gen['Amount'])
plt.title("Total Amount by Gender")
plt.show()
```

```python
# From above graphs we can see that most of the buyers are females and
# even the purchasing power of females are greater than males
```

```python
# Age :-
sns.countplot(x='Age Group', data = df, hue = 'Gender')
plt.title("Gender Wise Age Group Distribution")
plt.show()
```

```python
Sales_Age = df.groupby('Age Group')['Amount'].sum().sort_values(ascending = False).reset_index()
Sales_Age
```

```python
sns.set(rc = {'figure.figsize': (8,5)})
sns.barplot(data = Sales_Age, x = 'Age Group', y = 'Amount')
plt.title("Sales by Age Group")
plt.show()
```

```python
# From above graphs we can see that the most buyers are age group between 26-35 yrs females
```

```python
# State
Orders_State = df.groupby('State')['Orders'].sum().sort_values(ascending = False).head(10).reset_index()
Orders_State
```

```python
plt.figure(figsize = (20,5))
plt.bar(Orders_State['State'],Orders_State['Orders'])
plt.title("Top 10 States by Orders")
plt.show()
```

```python
Sales_State = df.groupby('State')['Amount'].sum().sort_values(ascending = False).head(10).reset_index()
Sales_State
```

```python
from matplotlib import figure
sns.set(rc = {'figure.figsize': (25,5)})
sns.barplot(data = Sales_State, x = 'State', y = 'Amount')
plt.title("Top 10 States by Sales")
plt.show()
```

```python
# From above graphs we can  see that most of the orders and sales/amount are
# from uttar pradesh, maharashtra and karnataka respectively.
```

```python
# Marital Status
sns.set(rc = {'figure.figsize': (3,5)})
sns.countplot(data = df, x = 'Marital_Status')
plt.title("Marital Status Distribution")
plt.show()
```

```python
Sales_Status = df.groupby(['Marital_Status','Gender'])['Amount'].sum().sort_values(ascending = False).reset_index()
Sales_Status
```

```python
sns.set(rc = {'figure.figsize': (6,5)})
sns.barplot(data = Sales_Status, x = 'Marital_Status', y = 'Amount', hue = 'Gender')
plt.title("Sales by Marital Status")
plt.show()
```

```python
# From above graphs we can  see that most of the buyers are married(women) and they have high purchasing power
```

```python
# occupation :-
sns.set(rc = {'figure.figsize': (20,8)})
sns.countplot(data = df, x = 'Occupation')
plt.title("Occupation Distribution")
plt.show()
```

```python
Sales_occupation = df.groupby('Occupation')['Amount'].sum().sort_values(ascending = False).reset_index()
Sales_occupation
```

```python
plt.bar(Sales_occupation['Occupation'], Sales_occupation['Amount'])
plt.title("Occupation Wise Sales")
plt.show()
```

```python
# From above graphs we can see that most of the buyers are working in IT, Healtcare and Aviation sector.
```

```python
# Product Category :-
sns.set(rc = {'figure.figsize': (30,12)})
plt.figure()
PC = sns.histplot(df['Product_Category'], kde=True)
PC.bar_label(PC.containers[0])
plt.title("Distribution of Product Category")
plt.show()
```

```python
Sales_PC = df.groupby('Product_Category')['Amount'].sum().sort_values(ascending = False).head(10).reset_index()
Sales_PC
```

```python
sns.set(rc = {'figure.figsize': (22,6)})
sns.barplot(data = Sales_PC, x = 'Product_Category', y = 'Amount')
plt.title("Sales by Product Category")
plt.show()
```

```python
# Frome above graphs we can see that most of the sold products are Food, Clothing and Electronics Category.
```

```python
# Top-10 Products :-
Sales_Top10 = df.groupby('Product_ID')['Orders'].sum().sort_values(ascending = False).head(10).reset_index()
Sales_Top10
```

```python
plt.bar(Sales_Top10['Product_ID'],Sales_Top10['Orders'])
plt.title("Top 10 Products by Orders")
plt.show()
```

```python
# Few products dominate sales.
```

```python
# Overall Correlation :-
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python
# no one is correlated.
```
